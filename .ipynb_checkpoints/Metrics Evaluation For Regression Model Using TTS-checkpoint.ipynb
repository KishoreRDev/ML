```python
import pandas as pd
```

```python
cars_data = pd.read_csv("Cars.csv")
cars_data
```

```python
cars_data.shape
```

```python
cars_data.isna().sum()
```

```python
cars_data.dtypes
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# Data seperation INPUT(X) AND OUTPUT(Y)
X = cars_data[["HP", "VOL", "SP", "WT"]]
X
```

```python
Y = cars_data.MPG
Y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size = 0.20, random_state = 42,shuffle = True )
X_train
```

```python
linear_model = LinearRegression()
```

```python
linear_model.fit(X_train, y_train)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

```python
y_pred_train = linear_model.predict(X_train)
y_pred_train
```

```python
y_pred_test = linear_model.predict(X_test)
y_pred_test
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Actual": y_train,
    "Predicted": y_pred_train
})

print(comparison.head(10))
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred_test
})

print(comparison.head(10))
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mean_absolute_error(y_train, y_pred_train)
```

```python
mean_squared_error(y_train, y_pred_train)
```

```python
r2_score(y_train, y_pred_train)
```

```python
import statsmodels.formula.api as smf
```

```python
model = smf.ols(formula="MPG ~ HP + VOL + SP + WT", data=cars_data).fit()  #Ordinary least Squares

print(model.summary())
```

```python
linear_stats_model = smf.ols(formula= "MPG ~ HP", data=cars_data).fit()
```

```python
print("R2 Scored = ",linear_stats_model.rsquared.round(4), "\nAdjusted R2 Scored", linear_stats_model.rsquared_adj.round(4))
```

```python
linear_stats_model = smf.ols(formula= "MPG ~ HP+VOL", data=cars_data).fit()
```

```python
print("R2 Scored = ",linear_stats_model.rsquared.round(4), "\nAdjusted R2 Scored", linear_stats_model.rsquared_adj.round(4))
```

```python
linear_stats_model = smf.ols(formula= "MPG ~ HP+VOL+SP", data=cars_data).fit()
```

```python
print("R2 Scored = ",linear_stats_model.rsquared.round(4), "\nAdjusted R2 Scored", linear_stats_model.rsquared_adj.round(4))
```

```python
linear_stats_model = smf.ols(formula= "MPG ~ HP+VOL+SP+WT", data=cars_data).fit()
```

```python
print("R2 Scored = ",linear_stats_model.rsquared.round(4), "\nAdjusted R2 Scored", linear_stats_model.rsquared_adj.round(4))
```

```python
import matplotlib.pyplot as plt

# Predict on test data
y_pred_test = linear_model.predict(X_test)

# Scatter plot
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred_test)
plt.xlabel("Actual MPG")
plt.ylabel("Predicted MPG")
plt.title("Actual vs Predicted")

# Perfect prediction line
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         color='red')

plt.show()
```

### R² Score: Shows how well the regression model explains the target variable. Higher is better, and the maximum value is 1.
### Adjusted R² Score: Like R², but it also checks whether the added features are actually useful. It penalizes unnecessary features.    
### This is why, in multiple linear regression, Adjusted R² is often preferred over R² when comparing models with different numbers of predictor variables.

R² says

"Your prediction is 92% good."

It does not care how many features you used.

Adjusted R² says

"Your prediction is 92% good, but you used too many unnecessary features."

It cares about the number of features.
